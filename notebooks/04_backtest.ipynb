# 04. Point-in-Time Look-Ahead Backtest & Placebo Analysis

This notebook evaluates the backtest simulation of the multi-factor scoring model
across 3 annual cohorts (FY2023, FY2024, FY2025) with a 3-month reporting lag to
prevent look-ahead bias.

```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd().resolve()
if not (project_root / "data").exists():
    project_root = project_root.parent

interim_dir = project_root / "data" / "interim"
```

## 1. Annual Cohort Performance by Quintile

Each annual cohort forms on June 30 (reflecting a 3-month lag following March 31
fiscal year-end) and holds for 12 months.

```python
cohort_path = interim_dir / "backtest_cohort_returns.csv"
if cohort_path.exists():
    df_cohort = pd.read_csv(cohort_path)
    print("Cohort 12-Month Holding Period Returns:")
    print(df_cohort.to_string(index=False))
```

## 2. Overall Backtest Performance Summary

Comparing cumulative and annualized performance of Quintiles 1 through 5 against
the Nifty 500 benchmark.

```python
summary_path = interim_dir / "backtest_summary.csv"
if summary_path.exists():
    df_summary = pd.read_csv(summary_path)
    print("Backtest Summary Statistics:")
    print(df_summary.to_string(index=False))
```

## 3. Placebo Comparison & Methodological Caveats

- The Top Quality Quintile (Q1) generated a 39.60% mean annual return, outperforming the
  benchmark in 3 of 3 periods (100% hit rate).
- The Bottom Quintile (Q5) generated 45.61% due to cyclical low-base rebounds in the
  2023-2024 bull market.
- In the 500-iteration placebo simulation, the random Q1 - Q5 spread 90% confidence interval
  spanned -11.55% to +11.56%. The observed spread of -6.01% falls within this random noise
  corridor, showing that short 3-year windows cannot statistically differentiate fundamental
  factor alpha from macroeconomic momentum.
