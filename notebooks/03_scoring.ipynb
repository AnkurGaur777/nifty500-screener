# 03. Sector-Relative Multi-Factor Scoring & Red-Flag Audit

This notebook evaluates the four-pillar multi-factor model, peer-group pooling,
cross-sectional Winsorization, red-flag triggers, and weight sensitivity.

```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd().resolve()
if not (project_root / "data").exists():
    project_root = project_root.parent

interim_dir = project_root / "data" / "interim"
```

## 1. Red-Flag Rule Trigger Rates

The red-flag engine evaluates 8 forensic accounting warning rules across all
non-financial companies.

```python
flags_path = interim_dir / "red_flags_summary.csv"
if flags_path.exists():
    df_flags = pd.read_csv(flags_path)
    print("Red-Flag Trigger Rates Summary:")
    print(df_flags.to_string(index=False))
```

## 2. Red-Flag Case Studies

Examining sample non-financial companies triggering forensic accounting rules.

```python
examples_path = interim_dir / "red_flags_examples.csv"
if examples_path.exists():
    df_examples = pd.read_csv(examples_path)
    print(f"Sample flagged cases ({len(df_examples)} records):")
    print(df_examples[["rule_name", "symbol", "name", "sector"]].head(10).to_string(index=False))
```

## 3. Weight Sensitivity Analysis

Assessing the stability of baseline top-20 leaders when testing against Quality,
Value, and Growth tilts.

```python
sens_path = interim_dir / "score_sensitivity.csv"
if sens_path.exists():
    df_sens = pd.read_csv(sens_path)
    print("Top-20 Overlap across Weighting Schemes:")
    print(df_sens.to_string(index=False))
```
